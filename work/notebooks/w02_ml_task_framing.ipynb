## 1. My lane as an ML task (type)

1. I selected the **Refresh / Content Opportunity Scoring** lane.

2. The **Refresh / Content Opportunity Scoring** lane is a **ranking/scoring** task. The goal is not to classify pages into fixed categories, but to rank content pages based on their potential benefit from being refreshed. The output is a prioritized list of pages that helps content editors decide which pages should be reviewed and updated first.

3. **WHY?** Content teams usually have limited time and resources, so they need to know **which pages should be updated first**, not just whether a page needs attention. A ranking approach produces a prioritized list or opportunity score, helping editors focus on the highest-impact pages before lower-priority ones.

## 2. Target or proxy

1. **Target**- The ideal target would be an **observed future outcome**, such as whether a content page shows improved performance (for example, higher clicks, impressions, or engagement) after being refreshed in a later time window. This type of target is measured from real outcomes rather than being created using a manually defined rule.

2. The starter dataset does not contain this future outcome, so this notebook focuses on defining the ML task and identifying a suitable target for future modeling rather than training a model with an artificial label.

## 3. Success metric

Since this is a ranking task, I would use **Precision@K** as the success metric. It measures how many of the top-ranked pages are actually good refresh opportunities.

A good result would mean that most of the pages recommended at the top of the list are useful for content editors to review and update, helping them spend their time on the pages with the highest potential impact.

## 4. The unit of analysis, as a real dataframe

The unit of analysis is **one content page (one content item)**.

Each row in the dataset represents a single content page and includes information such as its traffic, engagement, search performance, content characteristics, and refresh-related metrics over the past 90 days. The ranking model will assign a refresh opportunity score to each page so that editors can decide which pages should be updated first.

## 5. Why ML beats a fixed rule here

A fixed rule like "refresh pages older than 180 days" or "refresh pages with the lowest CTR" is not enough because a page's performance depends on many factors working together. Metrics such as impressions, clicks, engagement rate, search position, content age, and freshness all influence whether a page is a good refresh candidate.

An ML model can consider all these signals at the same time and rank pages based on their overall refresh opportunity. This gives editors better recommendations than relying on a single rule or threshold.
